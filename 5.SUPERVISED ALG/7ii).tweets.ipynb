# COLUMN TRANSFORMER

```python
preprocessing = ColumnTransformer(
        transformers=[
                ("count_vectorizer",
                 CountVectorizer(
                        stop_words="english",
                        lowercase=True
                        ),
                 "text")
                ],
        remainder="drop"
)
```

#  PIPELINE

```python
main_pipeline = Pipeline(
            steps=[
                ("preprocessing",preprocessing),
                ("model",SVC())
            ]
)
```

```python
grid_search_cv = GridSearchCV(
            estimator=main_pipeline,
            param_grid={
                "model__C":[0.1,1,10],
                "model__kernel":[
                            "linear",
                            "rbf"
                        ]},
            cv=5,
            verbose=1,
            n_jobs=-1
)
```

```python
grid_search_cv.fit(xtrain,ytrain)
```

```python
grid_search_cv.best_params_
```

```python
grid_search_cv.best_estimator_
```

```python
ypred = grid_search_cv.predict(xtest)
```

```python
print(classification_report(ytest,ypred))
```
