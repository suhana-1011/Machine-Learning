```python
import pandas as pd
from sklearn.cluster import KMeans
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler,OneHotEncoder
```

```python
df = pd.read_csv('Mall_Customers.csv')
```

```python
df.head()
```

```python
num_cols = df.select_dtypes(include='number').columns
```

```python
preprocessing = ColumnTransformer(
    transformers=[
        ('scaling',StandardScaler(),num_cols),
        ('encoder',OneHotEncoder(handle_unknown='ignore'),['Gender'])
    ]
)
main_pipeline = Pipeline(
    steps=[
        ('pre',preprocessing),
        ('model',KMeans(n_clusters=2,random_state=42))
    ]
)
main_pipeline.fit(df)
```

### Elbow method

```python
import matplotlib.pyplot as plt
metrics_ = []
for cluster in range(1,10):
    main_pipeline = Pipeline(
        steps=[
            ('pre',preprocessing),
            ('model',KMeans(n_clusters=cluster,random_state=42)) #8
        ]
    )    
    main_pipeline.fit(df)
    metric = main_pipeline.named_steps['model'].inertia_
    metrics_.append(metric)
    # print(main_pipeline.inertia_)
print(metrics_)
```

```python
plt.plot(range(1,10),metrics_,marker='o')
```
